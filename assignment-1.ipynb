```python
# Assignment 1: Operations on List, Tuple and Dictionary

# ---------------- LIST ----------------
print("=== LIST OPERATIONS ===")
fruits = ["apple", "banana", "cherry"]
print("Original list      :", fruits)

fruits.append("mango")                  # add at end
print("After append       :", fruits)

fruits.insert(1, "orange")              # add at index
print("After insert       :", fruits)

fruits.remove("banana")                 # remove by value
print("After remove       :", fruits)

popped = fruits.pop()                   # remove last item
print("Popped item        :", popped, "->", fruits)

fruits.extend(["grape", "kiwi"])        # add many items
print("After extend       :", fruits)

fruits.sort()                           # sort ascending
print("After sort         :", fruits)

fruits.reverse()                        # reverse order
print("After reverse      :", fruits)

print("Index of 'kiwi'    :", fruits.index("kiwi"))
print("Count of 'apple'   :", fruits.count("apple"))

# ---------------- TUPLE ----------------
print("\n=== TUPLE OPERATIONS ===")
nums = (10, 20, 30, 20, 40)
print("Tuple              :", nums)
print("First element      :", nums[0])
print("Slice [1:4]        :", nums[1:4])
print("Length             :", len(nums))
print("Count of 20        :", nums.count(20))
print("Index of 30        :", nums.index(30))
print("Max / Min          :", max(nums), "/", min(nums))
print("Sum                :", sum(nums))

# Tuples are immutable: 'append' and 'remove' are done by creating a new tuple
nums = nums + (50,)                     # append
print("After 'append' 50  :", nums)

temp = list(nums)                       # remove: tuple -> list -> tuple
temp.remove(20)
nums = tuple(temp)
print("After 'remove' 20  :", nums)

try:
    nums[0] = 99
except TypeError as e:
    print("Item assignment  : TypeError ->", e)

del nums                                # deletes the whole tuple

# ---------------- DICTIONARY ----------------
print("\n=== DICTIONARY OPERATIONS ===")
student = {"name": "Amit", "age": 20, "course": "Python"}
print("Dictionary         :", student)

student["city"] = "Pune"                # add a key-value pair
print("After adding city  :", student)

student["age"] = 21                     # update a value
print("After update age   :", student)

print("get('name')        :", student.get("name"))
print("get('phone')       :", student.get("phone", "Not found"))
print("keys()             :", list(student.keys()))
print("values()           :", list(student.values()))
print("items()            :", list(student.items()))

student.update({"grade": "A", "age": 22})   # add/update many
print("After update()     :", student)

removed = student.pop("city")           # remove by key
print("pop('city')        :", removed, "->", student)

last = student.popitem()                # remove last inserted pair
print("popitem()          :", last, "->", student)

del student["course"]                   # delete a key
print("After del course   :", student)

student.clear()                         # empty the dictionary
print("After clear()      :", student)
```
